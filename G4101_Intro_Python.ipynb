## f-Strings

f-Strings permiten crear cadenas de texto (strings) que incluyen valores de variables o expresiones dentro de llaves `{}`. Para crear un f-String se usa la letra `f` antes de las comillas que delimitan la cadena.

Para cambiar el formato de los valores dentro de las llaves se pueden usar especificadores de formato despues de dos puntos `:`. Por ejemplo, para mostrar un numero con 2 decimales se usa `{variable:.2f}`.

```python
name = "Ana"
age = 28
greeting = f"Hola, mi nombre es {name} y tengo {age} años."
print(greeting)
```

```python
pi = 3.14159265359
formatted_pi = f"El valor de pi con 2 decimales es {pi:.2f}."
print(formatted_pi)
```

Otra forma de formatear cadenas es usando el metodo `.format()`, donde los valores a insertar se pasan como argumentos al metodo, en el orden en que aparecen. Los especificadores de formato funcionan de la misma manera.

```python
temperatura = 23.56789
ciudad = "San Jose"
formatted_temp1 = "La temperatura actual en {} es {:.1f}°C".format(ciudad,temperatura)
formatted_temp2 = f"La temperatura actual en {ciudad} es {temperatura:.1f}°C."
```

```python
print(formatted_temp1)
```

```python
print(formatted_temp2)
```

## Funciones Lambda 

Funciones Lambda son funciones anonimas que se pueden definir en una sola linea de codigo. Estas son utiles para operaciones simples y rapidas, especialmente cuando se usan como argumentos en otras funciones, cuando no se necesita reutilizar la funcion, o para operaciones cortas.

```python
square = lambda x: x ** 2
print(square(5))
```

```python

add = lambda x, y: x + y
print(add(3, 7))
```

```python
multiply = lambda x, y: x * y
print(multiply(4, 6))
```

## Listas

En **Python** las listas se indexan empezando en 0, a diferencia de **R** donde se empieza en 1. Para acceder a los elementos de una lista se usan corchetes cuadrados `[]`, y para acceder a elementos dentro de elementos (sublistas) se usan multiples corchetes.

```python
lst = [x, y, "Hola", [1,2,3], np.array([4,5,6])]

print(lst[0])
```

## Vectores

Para trabajar con vectores en **Python** se usa el paquete **numpy**, el cual permite crear y manipular vectores de manera eficiente. En general se crean los arreglos con la función `np.array()`. **Python** y **numpy** diferencian entre vectores numericos (float) y enteros (int), mientras que en **R** no hay tal distinción. Si no se asignan de forma explicita los datos numericos se asignan como enteros si no tienen decimales y como float si tienen decimales.

Los vectores son unidimensionales, las matrices son bidimensionales, y los tensores son multidimensionales. En general los elementos deben ser del mismo tipo.

Se pueden crear vectores a partir de datos puntuales, los cuales pueden ser manipulados con operaciones aritmeticas u obtener estadisticas descriptivas.

```python
x = np.array([1,2,3,4,5], dtype=float)
y = np.array([6,7,8,9,10])

print(x + y)
```

Ejemplos de encadenamiento (chaining) de operaciones comunes usando **pandas** y **polars**.

```python
# Pandas: filtrar Temp>80 y Wind<=10, crear columna, agrupar y ordenar
result_pd = (
    airquality
    .loc[(airquality["Temp"] > 80) & (airquality["Wind"] <= 10)]
    .assign(Temp_C=lambda df: (df["Temp"] - 32) * 5/9)      # nueva columna derivada
    .groupby("Month", as_index=False)
    .agg(mean_temp=("Temp", "mean"), mean_wind=("Wind", "mean"))
    .sort_values("mean_temp", ascending=False)
)
result_pd
```

```python
# Polars: filtrar, crear columna y agregar por mes
result_pl = (
    air_pl
    .filter((pl.col("Temp") > 80) & (pl.col("Wind") <= 10))
    .with_columns(((pl.col("Temp") - 32) * 5/9).alias("Temp_C"))
    .group_by("Month")
    .agg([
        pl.col("Temp").mean().alias("mean_temp"),
        pl.col("Wind").mean().alias("mean_wind")
    ])
    .sort("mean_temp", descending=True)
)
result_pl
```

La libreria base para graficos en **Python** es **matplotlib**, la cual es muy poderosa pero su sintaxis puede ser complicada. Por esto, para graficos sencillos y estadisticos, se recomienda usar **seaborn**, la cual esta basada en **matplotlib** pero tiene una sintaxis mas sencilla y produce graficos mas atractivos por defecto.

De forma general se usan las funciones `plt.figure()` o `plt.subplots()` para crear la figura (`fig`) y los ejes (`ax`, `axes`), y despues se usan metodos de los ejes para agregar geoms, etiquetas, titulos, leyendas, etc. Al final se usa `plt.show()` para desplegar el grafico (opcional en Jupyter/Quarto).

Ejemplo del uso de **matplotlib**:

```python
np.random.seed(0)
x = np.linspace(0, 10, 100)
y = np.sin(x) + 0.12 * np.random.randn(100)
y2 = np.cos(x)

cats = ["A", "B", "C", "D"]
vals = [23, 45, 12, 30]

box_data = [np.random.normal(loc=i, scale=1, size=60) for i in range(4)]

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Línea + dispersión (superposición de geoms)
ax = axes[0, 0]
ax.plot(x, y2, label="cos(x)", color="C1", linewidth=2)
ax.scatter(x, y, label="sin(x) + ruido", color="C0", s=20, alpha=0.7)
ax.set_title("Línea + Dispersión")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()
ax.grid(True)

# Histograma
ax = axes[0, 1]
ax.hist(y, bins=20, color="C2", edgecolor="k")
ax.set_title("Histograma")
ax.set_xlabel("Valor")
ax.set_ylabel("Frecuencia")
ax.grid(axis="y", alpha=0.3)

# Barras categóricas
ax = axes[1, 0]
ax.bar(cats, vals, color="C3", edgecolor="k")
ax.set_title("Barras")
ax.set_xlabel("Categoría")
ax.set_ylabel("Valor")
for i, v in enumerate(vals):
  ax.annotate(str(v), (i, v), ha="center", va="bottom")

# Boxplot
ax = axes[1, 1]
ax.boxplot(box_data, labels=[f"Grp{i}" for i in range(1, 5)], patch_artist=True,
       boxprops=dict(facecolor="C4", color="k"))
ax.set_title("Boxplot")
ax.set_ylabel("Valor")

fig.tight_layout()
# fig.savefig("figs/matplotlib_structure.png", dpi=300)  # opción: guardar en archivo
fig
```

## plotnine

El paquete **plotnine** es el homologo a **ggplot2** de **R**. Este se basa en la gramatica de graficos (grammar of graphics). 

La estructura basica de cualquier grafico es:

```python
#| eval: false

ggplot(data = <DATA>) +
  <GEOM_FUNCTION>(mapping = aes(<MAPPINGS>),
                  stat = <STAT>, position = <POSITION>) +
  <FACET_FUNCTION> +
  <SCALE_FUNCTION> +
  <THEME_FUNCTION>
```

donde las partes principales son `ggplot` y `geom_*`, el resto no son necesarias. Dentro de `geom_*` se mapean la variables del DataFrame a los argumentos de la funcion (x, y, col, fill, size, shape).
